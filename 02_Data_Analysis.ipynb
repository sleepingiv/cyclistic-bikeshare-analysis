```python
import pandas as pd
df = pd.read_csv(r'Data\02_Prepared_Data\cleaned_cyclistic_data.csv')
```

```python
#To keep days of the week in chronological order instead of alphabetical
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['day_of_week'] = pd.Categorical(df['day_of_week'], categories=day_order, ordered=True)

#To keep months in chronological order
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
df['month'] = pd.Categorical(df['month'], categories=month_order, ordered=True)
```

```python
# Total number of rides by user type
total_rides = df['member_casual'].value_counts()
print('----Total Rides----')
print(total_rides)

# Average ride length by user type
avg_ride_length = df.groupby('member_casual')['ride_length'].mean()
print('\n ----Average ride length by user type----')
print(avg_ride_length)
```

```python
# Group by user type and day of the week to see the number of rides
rides_per_day = df.groupby(['member_casual','day_of_week'], observed = False).size().unstack()
print('-----Total number of rides per user type and day of the week-----')
print(rides_per_day)

# Group by user type and day of the week to see the average ride duration
avg_duration_per_day = df.groupby(['member_casual','day_of_week'], observed = False)['ride_length'].mean().unstack
print('\n-----Average duration of rides per user type and day of the week-----')
print(avg_duration_per_day)
```

```python
# Group by user type and month to observe seasonality
rides_by_month = df.groupby(['member_casual','month'], observed = False).size().unstack()
print('-----Total number of rides by user type and month of the year to notice seasonal changes-----')
print(rides_by_month)
```
